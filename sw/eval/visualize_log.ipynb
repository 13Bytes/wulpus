### One Channel

```python
from wulpus.plot_helpers import format_time_xticks


def make_measurement(row):
    idx_val = row.name
    time_val = row.get("time", 0)
    if np.isscalar(idx_val) and np.isfinite(idx_val):
        time_val = idx_val
    return {
        "data": row.get("measurement", []),
        "rx": row.get("rx", []),
        "tx": row.get("tx", []),
        "time": int(time_val),
        "mesh_origin": int(row.get("mesh_origin", 0)),
    }


def peak_metrics(arr: np.ndarray, ignore_n: int):
    arr = np.asarray(arr, dtype=float)
    if arr.ndim != 2:
        return np.array([]), np.array([]), np.array([]), arr, 0
    offset = ignore_n if arr.shape[1] > ignore_n else 0
    sliced = arr[:, offset:] if offset else arr
    if sliced.size == 0:
        return np.zeros(arr.shape[0]), np.zeros(arr.shape[0]), np.zeros(arr.shape[0]), sliced, offset
    p = np.nanmax(np.abs(sliced), axis=1)
    p2p = np.nanmax(sliced, axis=1) - np.nanmin(sliced, axis=1)
    abs_vals = np.abs(sliced)
    abs_vals[np.isnan(abs_vals)] = -np.inf
    peak_pos = np.argmax(abs_vals, axis=1) + offset
    peak_neg = np.argmin(sliced, axis=1) + offset
    return p, p2p, peak_pos, peak_neg, sliced, offset


def draw_peak_panel(ax_main, x, boundary_x, p, p2p, peak_pos, peak_neg, y_label, legend_labels, legend_y_offset=-0.5):
    for bx in boundary_x:
        ax_main.axvline(bx, color='grey', alpha=0.3, linewidth=1, zorder=0)
    (line_p,) = ax_main.plot(x, p, linewidth=1, label=legend_labels[0], zorder=3)
    (line_amp,) = ax_main.plot(x, p2p, linewidth=1, label=legend_labels[1], zorder=3)
    ax_main.set_ylabel(y_label)
    ax_main.set_xlabel('Time')
    ax_main.grid(True, alpha=0.1)
    ax_main.tick_params("x", rotation=90)
    ax_pos = ax_main.twinx()
    (line_pos,) = ax_pos.plot(x, peak_pos, linewidth=1, color='green', label=legend_labels[2], alpha=0.4, zorder=1)
    (line_neg,) = ax_pos.plot(x, peak_neg, linewidth=1, color='red', label=legend_labels[3], alpha=0.4, zorder=1)
    ax_pos.set_ylabel('Peak x-index')
    ax_main.legend(handles=[line_p, line_amp, line_pos, line_neg], loc='upper center', bbox_to_anchor=(0.5, legend_y_offset), fancybox=True, ncol=5)
    return ax_pos


# Plot with real time on x-axis from DataFrame index (helper-based)
fig, axs = plt.subplots(
    nrows=4, ncols=1, figsize=(11, 12), sharex=True,
    gridspec_kw={'height_ratios': [4, 1, 4, 1]}
 )
ax_img, ax_peak, ax_img_filt, ax_peak_filt = axs
border = 1000


# Select a single RX channel (0) and flatten to measurement columns
# df_channel_sel = df[df['rx'] == 0]
df_channel_sel = df.loc[df["rx"].apply(lambda rx: 5 in rx) & (df["wulpus"] == "E")]
if df_channel_sel.empty:
    raise ValueError("No data found for the selected RX channel and Wulpus ID.")



flattened_df, meas_cols = flatten_df_measurements(df_channel_sel, sample_crop=SAMPLE_CROP)
# Build (n_acq, n_samples) and transpose to (n_samples, n_acq) for imshow
values = flattened_df[meas_cols].to_numpy(dtype=float)  # (n_acq, n_samples)
plot_data = values.T  # (n_samples, n_acq)

# Use a single x for all bottom-plot traces (prevents stale-x shape mismatches)
x = np.arange(values.shape[0])


IGNORE_FIRST_N_SAMPLES = 12
processor = MeasurementProcessor()
filtered_rows = []
for _, row in df_channel_sel.iterrows():
    measurement = make_measurement(row)
    resp = processor.process_measurement(measurement=measurement, config=config)
    # IGNORE_FIRST_N_SAMPLES = int(resp.spacer_region[1])
    wavelet_env = np.asarray(resp.wavelet, dtype=float)
    original = np.asarray(row.get("measurement", []), dtype=float)
    if SAMPLE_CROP is not None:
        original = original[:SAMPLE_CROP]
    wavelet_resampled = resample(wavelet_env, original.size)
    filtered_rows.append(wavelet_resampled)

filtered_values = np.vstack(filtered_rows) if filtered_rows else np.empty_like(values)
filtered_plot_data = filtered_values.T if filtered_rows else np.empty((0, 0))



# Detect time discontinuities (stitched logs) to mark boundaries in the bottom plot
boundary_x = []
try:
    times = pd.to_datetime(df_channel_sel.index, errors='coerce')
    if hasattr(times, 'isna') and not times.isna().all():
        dt = pd.Series(times).diff().dt.total_seconds().to_numpy()
        dt_pos = dt[np.isfinite(dt) & (dt > 0)]
        typical_dt = float(np.median(dt_pos)) if dt_pos.size else float('nan')
        if np.isfinite(typical_dt) and typical_dt > 0:
            gap_threshold = 10.0 * typical_dt
            boundary_x = np.where((dt < 0) | (dt > gap_threshold))[0].tolist()
except Exception:
    boundary_x = []


# Choose a robust norm (symmetric range around 0 for visibility)
norm = colors.SymLogNorm(linthresh=10, linscale=0.03, vmin=-border, vmax=border)


# Top: heatmap (adds colorbar and local-time xticks)
imshow_with_time(
    ax_img, plot_data, index=df_channel_sel.index,
    cmap='viridis', interpolation='hamming', norm=norm, num_ticks=10, add_colorbar=False
 )
# Mark ignored samples line
ax_img.axhline(IGNORE_FIRST_N_SAMPLES, color='red', linestyle='--', linewidth=1)


ax_img.set_title('Measurement evolution over time (RX 0)')
ax_img.set_ylabel('Samples')
ax_img.grid(False)


# Raw peaks
legend_labels = ['Peak', 'Peak-to-peak', 'Pos peak [x]', 'Neg peak [x]']
p_raw, p2p_raw, peak_pos_raw, peak_neg_raw, values_peaks, _ = peak_metrics(values, IGNORE_FIRST_N_SAMPLES)
ax_peak_pos = draw_peak_panel(
    ax_peak, x, boundary_x, p_raw, p2p_raw, peak_pos_raw, peak_neg_raw,
    y_label='Peak-to-peak',
    legend_labels=legend_labels,
    legend_y_offset=-0.5,
 )


imshow_with_time(
    ax_img_filt, filtered_plot_data, index=df_channel_sel.index,
    cmap='viridis', interpolation='hamming', norm=norm, num_ticks=10, add_colorbar=False
 )
# Mark ignored samples line
ax_img_filt.axhline(IGNORE_FIRST_N_SAMPLES, color='red', linestyle='--', linewidth=1)
ax_img_filt.set_title('Filtered (wavelet envelope) evolution over time (RX 0)')
ax_img_filt.set_ylabel('Samples')
ax_img_filt.grid(False)


p_filt, p2p_filt, peak_pos_filt, peak_neg_filt, filtered_values_peaks, _ = peak_metrics(filtered_values, IGNORE_FIRST_N_SAMPLES)
ax_peak_pos_filt = draw_peak_panel(
    ax_peak_filt, x, boundary_x, p_filt, p2p_filt, peak_pos_filt, peak_neg_filt,
    y_label='Peak-to-peak',
    legend_labels=legend_labels,
    legend_y_offset=-1,
 )

plt.tight_layout()
plt.show()
```

```python
from wulpus.plot_helpers import format_time_index_to_local


def limit_slew_rate(data, max_change):
    limited_data = np.zeros_like(data)
    limited_data[0] = 0
    for i in range(1, len(data)):
        change = data[i] - limited_data[i - 1]
        limited_change = np.clip(change, -max_change, max_change)
        limited_data[i] = limited_data[i - 1] + limited_change
    return limited_data


x = x
peak_signal = p2p_filt  # can swap to p_raw if needed

peak_smooth = savgol_filter(peak_signal, 7, 2)
smooth_deriv = savgol_filter(peak_signal, 7, 2, deriv=1)

# Gradient of smoothed signal
grad_median = medfilt(smooth_deriv, kernel_size=29)

# Slew Rate Limiting
max_allowed_jump = 0.001 * np.std(peak_signal)
grad_slew_rate = limit_slew_rate(smooth_deriv, max_allowed_jump)

curing = np.where(grad_slew_rate <= -0.3)[0]
curing_idx = int(curing[0]) if curing.size else None

fig, axes = plt.subplots(2, 1, figsize=(10, 7), sharex=True)
axes[0].plot(x, peak_signal, alpha=0.4, linewidth=1, label='raw')
axes[0].plot(x, peak_smooth, linewidth=1.5, label='smoothed (Savitzky-Golay)')
axes[0].set_ylabel('Peak amplitude')
axes[0].set_title('Peak amplitude over time')
axes[0].grid(True, alpha=0.3)
axes[0].legend()

axes[1].plot(x, smooth_deriv, color='red', linewidth=1.2, label='gradient (Savitzky-Golay)', alpha=0.5)
axes[1].plot(x, grad_median, color='orange', linewidth=1.2, label='filtered gradient (Median filter)', alpha=0.8)
axes[1].plot(x, grad_slew_rate, color='green', linewidth=1.2, label='slew rate limited gradient', alpha=0.5)
axes[1].axhline(0, color='grey', linewidth=0.8, alpha=0.7)
axes[1].set_xlabel('Acquisition index')
axes[1].set_ylabel('d(peak)/dx')
axes[1].set_title('Gradient of smoothed peak amplitude')
axes[1].grid(True, alpha=0.3)
axes[1].legend()

if curing_idx is not None:
    axes[0].axvline(curing_idx, color='red', linestyle='--', linewidth=1, alpha=0.4)
    axes[1].axvline(curing_idx, color='red', linestyle='--', linewidth=1, alpha=0.4)

    idx_vals = group_df.index.values
    if np.issubdtype(idx_vals.dtype, np.integer):
        dt_index = pd.to_datetime(idx_vals, unit='us', errors='coerce')
    else:
        dt_index = pd.to_datetime(group_df.index, errors='coerce')

    dt_index = format_time_index_to_local(dt_index)
    print(f"First curing idx: {curing_idx}, timestamp: {dt_index[curing_idx].strftime('%H:%M:%S')}")


format_time_xticks(axes[1], index=group_df.index, num_ticks=10,
                       rotation=90)

plt.tight_layout()
plt.show()
```
